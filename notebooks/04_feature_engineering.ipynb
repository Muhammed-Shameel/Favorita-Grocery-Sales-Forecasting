```python
from catboost import CatBoostRegressor
import pandas as pd
import numpy as np
import time

# ============================================================
# 1. Recreate train / validation split
# ============================================================

VAL_START = pd.Timestamp("2017-07-01")
VAL_END   = pd.Timestamp("2017-07-16")

DROP_COLS = ["id", "sales", "transactions", "date"]

# Training = everything before validation
train_mask = df["date"] < VAL_START

# Validation = July 1 -> July 16
val_mask = (
    (df["date"] >= VAL_START) &
    (df["date"] <= VAL_END)
)

# Only select modelling columns
feature_cols = [
    col for col in df.columns
    if col not in DROP_COLS
]

X_train = df.loc[train_mask, feature_cols]
y_train = df.loc[train_mask, "sales"]

X_val = df.loc[val_mask, feature_cols]
y_val = df.loc[val_mask, "sales"]

print("Training rows:", len(X_train))
print("Validation rows:", len(X_val))
print("Features:", len(feature_cols))


# ============================================================
# 2. Take the same 500k training sample
# ============================================================

X_train_sample = X_train.sample(
    n=500_000,
    random_state=42
)

y_train_sample = y_train.loc[X_train_sample.index]

print("Sample:", X_train_sample.shape)


# ============================================================
# 3. Categorical features
# ============================================================

categorical_features = [
    "family",
    "city",
    "state",
    "type"
]

cat_feature_indices = [
    X_train_sample.columns.get_loc(col)
    for col in categorical_features
]


# ============================================================
# 4. Candidate configurations
# ============================================================

candidates = [
    {
        "name": "A_baseline",
        "depth": 8,
        "learning_rate": 0.05,
        "iterations": 700,
        "l2_leaf_reg": 3,
        "random_strength": 1.0,
        "bagging_temperature": 1.0,
    },
    {
        "name": "B_deeper",
        "depth": 9,
        "learning_rate": 0.04,
        "iterations": 850,
        "l2_leaf_reg": 5,
        "random_strength": 1.0,
        "bagging_temperature": 1.0,
    },
    {
        "name": "C_shallower",
        "depth": 7,
        "learning_rate": 0.04,
        "iterations": 900,
        "l2_leaf_reg": 5,
        "random_strength": 0.8,
        "bagging_temperature": 1.0,
    },
    {
        "name": "D_regularized",
        "depth": 8,
        "learning_rate": 0.04,
        "iterations": 900,
        "l2_leaf_reg": 10,
        "random_strength": 1.0,
        "bagging_temperature": 1.0,
    },
]


# ============================================================
# 5. Train candidates
# ============================================================

results = []
models = {}

for params in candidates:

    print("\n" + "=" * 70)
    print("TRAINING:", params["name"])
    print("=" * 70)

    start = time.time()

    model = CatBoostRegressor(
        iterations=params["iterations"],
        learning_rate=params["learning_rate"],
        depth=params["depth"],
        l2_leaf_reg=params["l2_leaf_reg"],
        random_strength=params["random_strength"],
        bagging_temperature=params["bagging_temperature"],

        loss_function="RMSE",
        eval_metric="MAE",

        random_seed=42,

        cat_features=cat_feature_indices,

        verbose=100,
        allow_writing_files=False,
        thread_count=4,

        od_type="Iter",
        od_wait=80,
        use_best_model=True,
    )

    model.fit(
        X_train_sample,
        y_train_sample,
        eval_set=(X_val, y_val),
    )

    predictions = model.predict(X_val)

    mae = np.mean(np.abs(y_val - predictions))
    rmse = np.sqrt(np.mean((y_val - predictions) ** 2))
    bias = np.mean(predictions - y_val)

    elapsed = time.time() - start

    results.append({
        "model": params["name"],
        "MAE": mae,
        "RMSE": rmse,
        "Bias": bias,
        "BestIteration": model.get_best_iteration(),
        "TimeMinutes": elapsed / 60,
    })

    models[params["name"]] = model

    print(f"\nMAE:            {mae:.4f}")
    print(f"RMSE:           {rmse:.4f}")
    print(f"Bias:           {bias:.4f}")
    print(f"Best iteration: {model.get_best_iteration()}")
    print(f"Time:            {elapsed / 60:.2f} min")


# ============================================================
# 6. Compare
# ============================================================

results_df = (
    pd.DataFrame(results)
    .sort_values("MAE")
    .reset_index(drop=True)
)

display(results_df)
```

```python
import numpy as np
import pandas as pd

from src.features.build_features import create_forecast_lag_features


# ============================================================
# 1. Settings
# ============================================================

VAL_START = pd.Timestamp("2017-07-01")
VAL_END = pd.Timestamp("2017-07-16")

model = models["B_deeper"]


# ============================================================
# 2. Keep ONLY columns needed for recursive forecasting
# ============================================================

needed_history_cols = [
    "date",
    "store_nbr",
    "family",
    "sales"
]

needed_forecast_cols = [
    col for col in model.feature_names_
    if col in df.columns
]

# create_forecast_lag_features also needs the group/date/sales
needed_forecast_cols = list(
    dict.fromkeys(
        ["date", "store_nbr", "family", "sales"]
        + needed_forecast_cols
    )
)


# ============================================================
# 3. Create history WITHOUT copying the entire dataframe
# ============================================================

history = df.loc[
    df["date"] < VAL_START,
    needed_history_cols
].copy()

validation = df.loc[
    (df["date"] >= VAL_START) &
    (df["date"] <= VAL_END),
    needed_forecast_cols
].copy()


# Sort once
history = history.sort_values(
    ["store_nbr", "family", "date"]
)

validation = validation.sort_values(
    ["date", "store_nbr", "family"]
).reset_index(drop=True)


print("History:", history.shape)
print("Validation:", validation.shape)


# ============================================================
# 4. Recursive prediction
# ============================================================

predictions = []

validation_dates = validation["date"].drop_duplicates()

for current_date in validation_dates:

    print(f"Processing {current_date.date()}...")

    # Current forecast day
    forecast_rows = validation.loc[
        validation["date"] == current_date
    ].copy()

    # --------------------------------------------------------
    # Create recursive lag features
    # --------------------------------------------------------

    forecast_features = create_forecast_lag_features(
        history,
        forecast_rows
    )

    # --------------------------------------------------------
    # Select exactly the features used during training
    # --------------------------------------------------------

    X_forecast = forecast_features[
        model.feature_names_
    ]

    # --------------------------------------------------------
    # Predict
    # --------------------------------------------------------

    day_predictions = model.predict(X_forecast)

    forecast_rows["prediction"] = day_predictions

    predictions.append(
        forecast_rows[
            [
                "date",
                "store_nbr",
                "family",
                "sales",
                "prediction"
            ]
        ]
    )

    # --------------------------------------------------------
    # Add predictions to history
    #
    # IMPORTANT:
    # We add predictions, NOT actual sales.
    # --------------------------------------------------------

    history_update = forecast_rows[
        [
            "date",
            "store_nbr",
            "family"
        ]
    ].copy()

    history_update["sales"] = day_predictions

    history = pd.concat(
        [
            history,
            history_update
        ],
        ignore_index=True
    )

    # Keep only the last 56 observations per store/family
    # because our largest lag is 56.
    history = (
        history
        .sort_values(
            ["store_nbr", "family", "date"]
        )
        .groupby(
            ["store_nbr", "family"],
            sort=False
        )
        .tail(56)
        .reset_index(drop=True)
    )


# ============================================================
# 5. Combine results
# ============================================================

recursive_results = pd.concat(
    predictions,
    ignore_index=True
)


# ============================================================
# 6. Metrics
# ============================================================

y_true = recursive_results["sales"].to_numpy()
y_pred = recursive_results["prediction"].to_numpy()

mae = np.mean(
    np.abs(y_true - y_pred)
)

rmse = np.sqrt(
    np.mean((y_true - y_pred) ** 2)
)

bias = np.mean(
    y_pred - y_true
)


# ============================================================
# 7. Results
# ============================================================

print("\n" + "=" * 60)
print("B_DEEPER — RECURSIVE EVALUATION")
print("=" * 60)

print(f"Rows: {len(recursive_results):,}")
print(f"MAE:  {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"Bias: {bias:.4f}")

print("\nCURRENT CHAMPION")
print("MAE:  65.2177")
print("RMSE: 230.0473")

print("\nIMPROVEMENT")

print(
    f"MAE improvement: "
    f"{(65.2177 - mae) / 65.2177 * 100:.2f}%"
)

print(
    f"RMSE improvement: "
    f"{(230.0473 - rmse) / 230.0473 * 100:.2f}%"
)
```

```python
# ============================================================
# E_depth9_lr035 — Recursive Evaluation
# ============================================================

VAL_START = pd.Timestamp("2017-07-01")
VAL_END = pd.Timestamp("2017-07-16")

# The model we are testing
model = models_2["E_depth9_lr035"]


# ============================================================
# 1. Only keep columns needed for recursive forecasting
# ============================================================

needed_history_cols = [
    "date",
    "store_nbr",
    "family",
    "sales"
]

needed_forecast_cols = list(
    dict.fromkeys(
        ["date", "store_nbr", "family", "sales"]
        + [
            col for col in model.feature_names_
            if col in df.columns
        ]
    )
)


# ============================================================
# 2. Create history and validation without copying all 53 cols
# ============================================================

history = df.loc[
    df["date"] < VAL_START,
    needed_history_cols
].copy()

validation = df.loc[
    (df["date"] >= VAL_START) &
    (df["date"] <= VAL_END),
    needed_forecast_cols
].copy()


history = history.sort_values(
    ["store_nbr", "family", "date"]
)

validation = validation.sort_values(
    ["date", "store_nbr", "family"]
).reset_index(drop=True)


print("History:", history.shape)
print("Validation:", validation.shape)


# ============================================================
# 3. Recursive forecasting
# ============================================================

predictions = []

for current_date in validation["date"].drop_duplicates():

    print(f"Processing {current_date.date()}...")

    # Current day
    forecast_rows = validation.loc[
        validation["date"] == current_date
    ].copy()

    # Build lag/rolling features from CURRENT history
    forecast_features = create_forecast_lag_features(
        history,
        forecast_rows
    )

    # Select exact model features
    X_forecast = forecast_features[
        model.feature_names_
    ]

    # Predict
    day_predictions = model.predict(X_forecast)

    forecast_rows["prediction"] = day_predictions

    predictions.append(
        forecast_rows[
            [
                "date",
                "store_nbr",
                "family",
                "sales",
                "prediction"
            ]
        ]
    )

    # --------------------------------------------------------
    # Add predictions to history
    # --------------------------------------------------------

    history_update = forecast_rows[
        [
            "date",
            "store_nbr",
            "family"
        ]
    ].copy()

    history_update["sales"] = day_predictions

    history = pd.concat(
        [
            history,
            history_update
        ],
        ignore_index=True
    )

    # Keep only the latest 56 observations per store/family
    history = (
        history
        .sort_values(
            ["store_nbr", "family", "date"]
        )
        .groupby(
            ["store_nbr", "family"],
            sort=False
        )
        .tail(56)
        .reset_index(drop=True)
    )


# ============================================================
# 4. Combine predictions
# ============================================================

recursive_results_E = pd.concat(
    predictions,
    ignore_index=True
)


# ============================================================
# 5. Metrics
# ============================================================

y_true = recursive_results_E["sales"].to_numpy()
y_pred = recursive_results_E["prediction"].to_numpy()

mae = np.mean(
    np.abs(y_true - y_pred)
)

rmse = np.sqrt(
    np.mean((y_true - y_pred) ** 2)
)

bias = np.mean(
    y_pred - y_true
)


# ============================================================
# 6. Compare with current champion
# ============================================================

champion_mae = 63.3448
champion_rmse = 226.2852

print("\n" + "=" * 60)
print("E_depth9_lr035 — RECURSIVE EVALUATION")
print("=" * 60)

print(f"Rows: {len(recursive_results_E):,}")
print(f"MAE:  {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"Bias: {bias:.4f}")

print("\nCURRENT CHAMPION")
print(f"MAE:  {champion_mae:.4f}")
print(f"RMSE: {champion_rmse:.4f}")

print("\nIMPROVEMENT")

print(
    f"MAE improvement: "
    f"{(champion_mae - mae) / champion_mae * 100:.2f}%"
)

print(
    f"RMSE improvement: "
    f"{(champion_rmse - rmse) / champion_rmse * 100:.2f}%"
)
```
