# Understanding the data relationship

## Understanding the Raw Datasets

### 1. Training Dataset (`train.csv`)
Each row represents the sales of a specific product family at a particular store on a specific date. It includes the sales quantity and the number of items on promotion.

### 2. Holidays and Events Dataset (`holidays_events.csv`)
Each row represents a holiday or event, including its date, type, location or geographical scope, description, and whether it was transferred to another date.

### 3. Oil Dataset (`oil.csv`)
Each row represents the daily oil price for a specific date. This dataset provides economic context that may help explain changes in retail sales.

### 4. Stores Dataset (`stores.csv`)
Each row represents a store and its metadata, including the store number, city, state, store type, and cluster. The store type (e.g., A, D) categorizes the store, while the cluster groups stores with similar characteristics.

### 5. Transactions Dataset (`transactions.csv`)
Each row represents the number of transactions recorded at a particular store on a specific date. This dataset provides additional information about store activity and customer purchasing patterns.
